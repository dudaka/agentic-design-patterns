# CrewAI

```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, Process, LLM

load_dotenv(override=True)
if not os.getenv("GOOGLE_API_KEY"):
    raise ValueError("GOOGLE_API_KEY not found. Please set it in your .env file.")

# CrewAI takes its own LLM object, not a LangChain chat model, and routes it through
# LiteLLM - hence the 'gemini/' prefix. The book's gemini-2.0-flash is retired.
llm = LLM(model="gemini/gemini-flash-latest")

# Define Agents with specific roles and goals
researcher = Agent(
    role='Senior Research Analyst',
    goal='Find and summarize the latest trends in AI.',
    backstory="You are an experienced research analyst with a knack for identifying key trends and synthesizing information.",
    verbose=True,
    allow_delegation=False,
    llm=llm,  # Assign the specific LLM to the agent
)

writer = Agent(
    role='Technical Content Writer',
    goal='Write a clear and engaging blog post based on research findings.',
    backstory="You are a skilled writer who can translate complex technical topics into accessible content.",
    verbose=True,
    allow_delegation=False,
    llm=llm,
)

# Define Tasks for the agents
research_task = Task(
    description="Research the top 3 emerging trends in Artificial Intelligence in 2024-2025. Focus on practical applications and potential impact.",
    expected_output="A detailed summary of the top 3 AI trends, including key points and sources.",
    agent=researcher,
)

writing_task = Task(
    description="Write a 500-word blog post based on the research findings. The post should be engaging and easy for a general audience to understand.",
    expected_output="A complete 500-word blog post about the latest AI trends.",
    agent=writer,
    context=[research_task],  # the writer receives the researcher's output
)

# Create the Crew. Crew has no llm field, so the model is assigned per agent above.
blog_creation_crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, writing_task],
    process=Process.sequential,
    verbose=True,  # bool in CrewAI 1.x; the book's verbose=2 raises a ValidationError
)

# Execute the Crew.
# The notebook already runs an event loop, so use the async kickoff.
print("## Running the blog creation crew with Gemini... ##")
result = await blog_creation_crew.kickoff_async()

print("\n------------------\n")
print("## Crew Final Output ##")
print(result)
```

## ADK Gemini AgentTool

```python
import uuid

from google.adk.agents import LlmAgent
from google.adk.runners import InMemoryRunner
from google.adk.tools import agent_tool
from google.genai import types

GEMINI_MODEL = "gemini-flash-latest"  # the book's gemini-2.0-flash is retired

# 1. A simple function tool for the core capability.
# This follows the best practice of separating actions from reasoning.
def generate_image(prompt: str) -> dict:
    """
    Generates an image based on a textual prompt.

    Args:
        prompt: A detailed description of the image to generate.

    Returns:
        A dictionary with the status and the generated image bytes.
    """
    print(f"TOOL: Generating image for prompt: '{prompt}'")
    # In a real implementation, this would call an image generation API.
    # For this example, we return mock image data.
    mock_image_bytes = b"mock_image_data_for_a_cat_wearing_a_hat"
    return {
        "status": "success",
        # The tool returns the raw bytes, the agent will handle the Part creation.
        "image_bytes": mock_image_bytes,
        "mime_type": "image/png"
    }


# 2. Refactor the ImageGeneratorAgent into an LlmAgent.
# It now correctly uses the input passed to it.
image_generator_agent = LlmAgent(
    name="ImageGen",
    model=GEMINI_MODEL,
    description="Generates an image based on a detailed text prompt.",
    instruction=(
        "You are an image generation specialist. Your task is to take the user's request "
        "and use the `generate_image` tool to create the image. "
        "The user's entire request should be used as the 'prompt' argument for the tool. "
        "After the tool returns the image bytes, you MUST output the image."
    ),
    tools=[generate_image]
)

# 3. Wrap the corrected agent in an AgentTool.
# AgentTool no longer takes a description: the parent agent sees the wrapped agent's
# own name and description, so ImageGen's description above is what the Artist reads.
image_tool = agent_tool.AgentTool(agent=image_generator_agent)

# 4. The parent agent remains unchanged. Its logic was correct.
artist_agent = LlmAgent(
    name="Artist",
    model=GEMINI_MODEL,
    instruction=(
        "You are a creative artist. First, invent a creative and descriptive prompt for an image. "
        "Then, use the `ImageGen` tool to generate the image using your prompt."
    ),
    tools=[image_tool]
)

# --- How it works now ---
# 1. The `artist_agent` decides on a prompt, e.g., "A photorealistic cat wearing a tiny top hat."
# 2. It calls the tool: `ImageGen(input="A photorealistic cat wearing a tiny top hat.")`
#    (Note: AgentTool uses 'input' as the default parameter name for the sub-agent's query).
# 3. The `agent_tool` invokes `image_generator_agent` with the prompt as its input.
# 4. The `image_generator_agent` follows its instructions and calls `generate_image(prompt="...")`.
# 5. The function returns the image bytes.
# 6. `image_generator_agent` returns the final image as its result.
# 7. `artist_agent` receives the image result from the tool call.

# 5. Execution. The book stops at the definitions, so drive the parent agent here.
runner = InMemoryRunner(artist_agent)

async def run_artist(request: str):
    """Runs the artist agent once and prints its final response."""
    print(f"\n--- Running Artist with request: '{request}' ---")
    user_id = "user_123"
    session_id = str(uuid.uuid4())
    # create_session is a coroutine and must be awaited, or the session is never created.
    await runner.session_service.create_session(
        app_name=runner.app_name, user_id=user_id, session_id=session_id
    )

    async for event in runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=types.Content(role='user', parts=[types.Part(text=request)]),
    ):
        if event.is_final_response() and event.content and event.content.parts:
            text = "".join(part.text for part in event.content.parts if part.text)
            # event.author names the agent that produced the response.
            print(f"[{event.author}] {text}")

# The notebook already runs an event loop, so await the coroutine directly.
await run_artist("Draw me something whimsical.")
```

### ADK Single-Turn Sub-Agent

The cell below does the same thing as the one above: `Artist` invents a prompt, calls `ImageGen`, which calls `generate_image`. Agent names, models, descriptions, instructions, the request, and the runner scaffolding are all identical. What differs is **how the sub-agent is attached to the parent**.

| | Above (the book's version) | Below (current ADK) |
|---|---|---|
| Attachment | `agent_tool.AgentTool(agent=...)` passed as `tools=[image_tool]` | `mode="single_turn"` on the child, then `sub_agents=[...]` |
| Who builds the tool | you do, explicitly | ADK does it for you |
| API status | discouraged by `AgentTool`'s own docstring | the recommended form |

There is no wrapper below. ADK scans `sub_agents`, sees `mode="single_turn"`, and appends an internal `_SingleTurnAgentTool` to the parent's tool list itself. That is why the cell prints `single_turn_artist_agent.tools` and gets `['ImageGen']` — a tool that nothing in the cell explicitly created.

### Why `mode` is the interesting part

`sub_agents` alone does **not** give you agent-as-tool. The default `mode` for a sub-agent is `chat`, and that is chapter 2's `transfer_to_agent` handoff: control moves to the child and the parent is done. `single_turn` is call-and-return — the child finishes its task, hands back a result, and the parent carries on.

So the cell below uses the *same keyword* as chapter 2's routing example with the opposite control flow. One field selects which collaboration pattern you get.

### Three practical differences

1. **The cell below cannot run on its own.** It reuses `uuid`, `types`, `LlmAgent`, `InMemoryRunner`, `GEMINI_MODEL`, and `generate_image` from the cell above — no imports, no tool definition. Run the cells in order; running it alone after a kernel restart raises `NameError`.
2. **The agents are re-declared** with `single_turn_` prefixes rather than reused. An ADK agent may have only one parent, so passing the cell above's `image_generator_agent` into `sub_agents` would fail.
3. **The book's seven-step "How it works now" comment block stays above**, because it describes the `AgentTool` path specifically.

### One difference worth not over-reading

Across test runs the cell below sometimes emitted several final-response events where the cell above emitted one. Do not treat that as a reliable distinction: the cell above also varied between runs, calling `generate_image` once in one run and twice in another, so some of this is model nondeterminism.

There is a documented mechanism that may underlie it. ADK's docstring says a `single_turn` sub-agent "runs inline in the parent's session", whereas `AgentTool` gives the wrapped agent its own runner (its `include_plugins` flag exists to propagate plugins into that separate runner). Inline versus isolated execution is a real architectural difference, but it has not been traced to the event counts above, so treat the connection as plausible rather than established.

```python
# The same two-agent example, written the way ADK now recommends.
# AgentTool's own docstring calls direct use discouraged: instead mark the sub-agent
# mode="single_turn" and attach it with sub_agents, and the framework exposes it as a
# tool automatically. Agents are re-declared because an agent may have only one parent.

single_turn_image_agent = LlmAgent(
    name="ImageGen",
    model=GEMINI_MODEL,
    description="Generates an image based on a detailed text prompt.",
    instruction=(
        "You are an image generation specialist. Your task is to take the user's request "
        "and use the `generate_image` tool to create the image. "
        "The user's entire request should be used as the 'prompt' argument for the tool. "
        "After the tool returns the image bytes, you MUST output the image."
    ),
    tools=[generate_image],
    # single_turn: completes its task and returns, without chatting with the user.
    # The default for a sub-agent is "chat", which is a transfer_to_agent handoff instead.
    mode="single_turn",
)

single_turn_artist_agent = LlmAgent(
    name="Artist",
    model=GEMINI_MODEL,
    instruction=(
        "You are a creative artist. First, invent a creative and descriptive prompt for an image. "
        "Then, use the `ImageGen` tool to generate the image using your prompt."
    ),
    # No AgentTool wrapper: ADK turns each single_turn sub-agent into a tool on this agent.
    sub_agents=[single_turn_image_agent],
)

single_turn_runner = InMemoryRunner(single_turn_artist_agent)

async def run_single_turn_artist(request: str):
    """Runs the sub_agents version of the artist agent and prints its final response."""
    print(f"\n--- Running Artist (single_turn sub-agent) with request: '{request}' ---")
    user_id = "user_123"
    session_id = str(uuid.uuid4())
    await single_turn_runner.session_service.create_session(
        app_name=single_turn_runner.app_name, user_id=user_id, session_id=session_id
    )

    async for event in single_turn_runner.run_async(
        user_id=user_id,
        session_id=session_id,
        new_message=types.Content(role='user', parts=[types.Part(text=request)]),
    ):
        if event.is_final_response() and event.content and event.content.parts:
            text = "".join(part.text for part in event.content.parts if part.text)
            # event.author names the agent that produced the response. ADK runs a
            # single_turn sub-agent inline in the parent's session, and in practice this
            # stream can carry more than one final-response event.
            print(f"[{event.author}] {text}")

# Confirm the sub-agent really did become a tool on the parent.
print("Artist tools:", [t.name for t in single_turn_artist_agent.tools])

await run_single_turn_artist("Draw me something whimsical.")
```

## ADK Gemini Coordinator

```python
from google.adk.agents import LlmAgent, BaseAgent
from google.adk.agents.invocation_context import InvocationContext
from google.adk.events import Event
from typing import AsyncGenerator

# Correctly implement a custom agent by extending BaseAgent
class TaskExecutor(BaseAgent):
    """A specialized agent with custom, non-LLM behavior."""
    name: str = "TaskExecutor"
    description: str = "Executes a predefined task."

    async def _run_async_impl(self, context: InvocationContext) -> AsyncGenerator[Event, None]:
        """Custom implementation logic for the task."""
        # This is where your custom logic would go.
        # For this example, we'll just yield a simple event.
        yield Event(author=self.name, content="Task finished successfully.")

# Define individual agents with proper initialization
# LlmAgent requires a model to be specified.
greeter = LlmAgent(
    name="Greeter",
    model="gemini-2.0-flash-exp",
    instruction="You are a friendly greeter."
)
task_doer = TaskExecutor() # Instantiate our concrete custom agent

# Create a parent agent and assign its sub-agents
# The parent agent's description and instructions should guide its delegation logic.
coordinator = LlmAgent(
    name="Coordinator",
    model="gemini-2.0-flash-exp",
    description="A coordinator that can greet users and execute tasks.",
    instruction="When asked to greet, delegate to the Greeter. When asked to perform a task, delegate to the TaskExecutor.",
    sub_agents=[
        greeter,
        task_doer
    ]
)

# The ADK framework automatically establishes the parent-child relationships.
# These assertions will pass if checked after initialization.
assert greeter.parent_agent == coordinator
assert task_doer.parent_agent == coordinator

print("Agent hierarchy created successfully.")
```

### ADK Gemini Coordinator, updated

The cell above prints `Agent hierarchy created successfully.` and looks healthy. It is not — it simply never runs the agents it builds. Constructing an agent validates very little, so two faults sit dormant:

| Fault | Why the cell above hides it |
|---|---|
| `model="gemini-2.0-flash-exp"` | Retired: `404 NOT_FOUND ... not found for API version v1beta`. A model name is only resolved on first *use*, and nothing here uses it. |
| `Event(content="Task finished successfully.")` | `Event.content` is `Optional[types.Content]`; a bare string raises a pydantic `ValidationError`. `_run_async_impl` is a generator body, so it never executed. |

This is a good moment to make a point about agentic code in general: **construction is not validation.** A cell that only wires objects together and asserts their relationships can pass while every path through it is broken. The assertions here check `parent_agent`, which ADK sets during construction — they were never going to fail.

The cell below fixes both faults and then *runs* the coordinator, so the delegation is demonstrated rather than asserted.

What did **not** need changing: subclassing `BaseAgent` and overriding `_run_async_impl(self, context: InvocationContext)` is still the current way to write a custom, non-LLM agent — ADK's own `LoopAgent` is written exactly that way. Only the `Event` it yields was wrong.

One connection worth drawing: `sub_agents` here is chapter 2's *delegation*, not the agent-as-tool of the earlier cells. The coordinator hands control off to `Greeter` or `TaskExecutor` and does not get it back, which is why `event.author` in the output names the sub-agent rather than the coordinator.

```python
# Same coordinator, brought up to date and actually executed.
#
# The cell above constructs the agents and asserts the hierarchy, but never runs them.
# That is what let two latent faults survive:
#   1. model="gemini-2.0-flash-exp" is retired and returns 404 on first use.
#   2. Event(content="Task finished successfully.") is invalid - Event.content must be a
#      types.Content, and a plain string raises a pydantic ValidationError. The body of
#      _run_async_impl never executed, so nothing ever raised it.

from typing import AsyncGenerator

from google.adk.agents import LlmAgent, BaseAgent
from google.adk.agents.invocation_context import InvocationContext
from google.adk.events import Event
from google.adk.runners import InMemoryRunner
from google.genai import types

GEMINI_MODEL = "gemini-flash-latest"


# A custom agent is still a BaseAgent subclass overriding _run_async_impl: that part of
# the book is current. Only the Event it yields has to change.
class TaskExecutor(BaseAgent):
    """A specialized agent with custom, non-LLM behavior."""

    name: str = "TaskExecutor"
    description: str = "Executes a predefined task."

    async def _run_async_impl(
        self, context: InvocationContext
    ) -> AsyncGenerator[Event, None]:
        """Custom implementation logic for the task."""
        # Event.content takes a types.Content, never a bare string.
        yield Event(
            author=self.name,
            content=types.Content(
                role="model",
                parts=[types.Part(text="Task finished successfully.")],
            ),
        )


greeter = LlmAgent(
    name="Greeter",
    model=GEMINI_MODEL,
    instruction="You are a friendly greeter.",
)
task_doer = TaskExecutor()

coordinator = LlmAgent(
    name="Coordinator",
    model=GEMINI_MODEL,
    description="A coordinator that can greet users and execute tasks.",
    instruction=(
        "When asked to greet, delegate to the Greeter. "
        "When asked to perform a task, delegate to the TaskExecutor."
    ),
    sub_agents=[greeter, task_doer],
)

# The ADK framework automatically establishes the parent-child relationships.
assert greeter.parent_agent is coordinator
assert task_doer.parent_agent is coordinator
print("Agent hierarchy created successfully.")

# --- Run it, so the delegation is demonstrated rather than asserted ---
APP_NAME = "coordinator_app"
USER_ID = "user_1"

runner = InMemoryRunner(agent=coordinator, app_name=APP_NAME)


async def ask(request: str):
    """Sends one request to the coordinator and prints who answered."""
    session = await runner.session_service.create_session(
        app_name=APP_NAME, user_id=USER_ID
    )
    print(f"\n--- {request} ---")
    async for event in runner.run_async(
        user_id=USER_ID,
        session_id=session.id,
        new_message=types.Content(role="user", parts=[types.Part(text=request)]),
    ):
        if event.is_final_response() and event.content and event.content.parts:
            text = "".join(part.text for part in event.content.parts if part.text)
            if text:
                # event.author shows which agent the coordinator handed off to.
                print(f"[{event.author}] {text}")


await ask("Say hello to me.")
await ask("Please perform the task.")
```

## ADK Gemini Loop

```python
import asyncio
from typing import AsyncGenerator
from google.adk.agents import LoopAgent, LlmAgent, BaseAgent
from google.adk.events import Event, EventActions
from google.adk.agents.invocation_context import InvocationContext

# Best Practice: Define custom agents as complete, self-describing classes.
class ConditionChecker(BaseAgent):
    """A custom agent that checks for a 'completed' status in the session state."""
    name: str = "ConditionChecker"
    description: str = "Checks if a process is complete and signals the loop to stop."

    async def _run_async_impl(
        self, context: InvocationContext
    ) -> AsyncGenerator[Event, None]:
        """Checks state and yields an event to either continue or stop the loop."""
        status = context.session.state.get("status", "pending")
        is_done = (status == "completed")

        if is_done:
            # Escalate to terminate the loop when the condition is met.
            yield Event(author=self.name, actions=EventActions(escalate=True))
        else:
            # Yield a simple event to continue the loop.
            yield Event(author=self.name, content="Condition not met, continuing loop.")

# Correction: The LlmAgent must have a model and clear instructions.
process_step = LlmAgent(
    name="ProcessingStep",
    model="gemini-2.0-flash-exp",
    instruction="You are a step in a longer process. Perform your task. If you are the final step, update session state by setting 'status' to 'completed'."
)

# The LoopAgent orchestrates the workflow.
poller = LoopAgent(
    name="StatusPoller",
    max_iterations=10,
    sub_agents=[
        process_step,
        ConditionChecker() # Instantiating the well-defined custom agent.
    ]
)

# This poller will now execute 'process_step' and then 'ConditionChecker'
# repeatedly until the status is 'completed' or 10 iterations have passed.
```

### ADK Gemini Loop with Workflow

The cell above raises `DeprecationWarning: LoopAgent is deprecated in favor of Workflow`. The cell below is the same loop with no deprecated API.

The translation is not one-to-one, which is the point worth teaching:

| `LoopAgent` (above) | `Workflow` (below) |
|---|---|
| `max_iterations=10` | no such field — the routing node counts passes itself |
| a `ConditionChecker` agent yielding `EventActions(escalate=True)` | a routing function that sets `ctx.route` |
| loop is implicit in the agent type | loop is an explicit **conditional edge that points backwards** |

`LoopAgent` gave you a loop and you supplied a way out of it. `Workflow` has no loop at all: a cycle is just an edge whose route maps back to an earlier node, so the routing node owns both the stop condition and the iteration cap. That also means a `Workflow` needs a terminal node to land on when the loop ends — hence `report`.

Two gotchas found while porting:

- A routing function must **set `ctx.route`**, not return an `Event(actions=EventActions(route=...))`. Returning the event silently fails: the graph logs `none were matched by the emitted route(s): None` and the branch simply ends after one pass.
- On loop-back the worker receives only the previous node's output, not the original request. The book's `ProcessingStep` was told to "set session state to completed" but had no way to do it, so that loop could only ever stop on `max_iterations`. The instruction below states the stop condition in terms of what the worker actually sees.

The remaining `BaseAgentConfig is deprecated` warning is internal to google-adk 2.7.1 — it fires on a bare `Agent(...)` construction and affects every ADK cell in this repo.

```python
# Same status-polling loop, written with the graph-based Workflow API (google-adk 2.x).
# LoopAgent above is deprecated: "LoopAgent is deprecated in favor of Workflow and will
# be removed in a future version."
#
# The translation is not one-to-one. A Workflow has no max_iterations and no escalate
# signal, so there is nothing for a ConditionChecker agent to escalate to. Instead a loop
# IS a conditional edge that routes back to an earlier node, and the routing node owns
# both the stop condition and the iteration cap.

from google.adk import Agent, Workflow
from google.adk.agents.context import Context
from google.adk.runners import InMemoryRunner
from google.genai import types

GEMINI_MODEL = "gemini-flash-latest"
MAX_PASSES = 10

# The worker node. One pass of the process per visit.
process_step = Agent(
    name="ProcessingStep",
    model=GEMINI_MODEL,
    instruction=(
        "You are one step in a counting process. You receive the previous step's output. "
        "Reply with just the next number in the count. "
        "If the previous step's number is 3 or higher, reply with exactly: COMPLETED"
    ),
)


def check_condition(node_input: str, ctx: Context) -> str:
    """Routes back to ProcessingStep until the worker reports COMPLETED, or the cap is hit."""
    passes = ctx.state.get("passes", 0) + 1
    ctx.state["passes"] = passes

    reported_done = "COMPLETED" in (node_input or "").upper()
    stop = reported_done or passes >= MAX_PASSES

    # Setting ctx.route is what the conditional edge below switches on.
    ctx.route = "done" if stop else "continue"
    print(f"  [check] pass {passes}: {ctx.route}")
    return node_input


def report(node_input: str, ctx: Context) -> str:
    """Terminal node: a Workflow needs somewhere to land when the loop ends."""
    summary = f"Loop finished after {ctx.state.get('passes', 0)} pass(es)."
    print(summary)
    return summary


# The cycle: check_condition routes back to process_step, or forward to report.
poller = Workflow(
    name="StatusPoller",
    edges=[
        ("START", process_step, check_condition),
        (check_condition, {"continue": process_step, "done": report}),
    ],
)

# --- Run the workflow ---
APP_NAME = "workflow_poller_app"
USER_ID = "user_1"

runner = InMemoryRunner(agent=poller, app_name=APP_NAME)
session = await runner.session_service.create_session(app_name=APP_NAME, user_id=USER_ID)

message = types.Content(
    role="user",
    parts=[types.Part(text="Start counting. The first number is 1.")],
)

async for event in runner.run_async(
    user_id=USER_ID, session_id=session.id, new_message=message
):
    if event.is_final_response() and event.content and event.content.parts:
        text = event.content.parts[0].text
        if text:
            print(f"--- {event.author} ---")
            print(text)
```

### ADK Gemini Parallel with Workflow

The cell above has three problems, and only the first is the deprecation:

| Problem | Detail |
|---|---|
| `ParallelAgent` is deprecated | in favour of `Workflow`, exactly like `LoopAgent` above |
| `model="gemini-2.0-flash-exp"` | retired — `404 NOT_FOUND` on first use |
| the `Runner` block is entirely commented out | nothing in it ever executes; the book offers it as a "conceptual example" |

That last one matters more than it looks. Because the runner is commented out, the retired model never gets resolved and the cell appears to succeed — the same trap as the coordinator cell above. **Construction is not validation.**

How the two APIs express parallelism:

| `ParallelAgent` (above) | `Workflow` (below) |
|---|---|
| `sub_agents=[a, b]` | a **tuple** inside one edge: `("START", (a, b), join, report)` |
| results land in `session.state` via `output_key` | a `JoinNode` hands the next node a dict **keyed by upstream node name** |

A `JoinNode` is a synchronization barrier: it waits for every incoming node to finish, then packages their outputs into one dict. That dict is what replaces `output_key` here — collecting fan-out results is a property of the graph rather than something each agent writes into shared state.

Worth pointing out in class: this is the same shape chapter 3 already uses for its parallel research pipeline, where three researchers fan out into a `JoinNode` and then into a merger agent. Chapter 3 fans out to *do* independent work; this chapter fans out as one shape of multi-agent collaboration. Same graph, different reason.

One caveat inherited from the book: these two agents have no tools, so the weather and the news are simply invented by the model. The book flags this itself — "in a real-world scenario, you would use tools."

```python
# Same parallel data gatherer, written with the graph-based Workflow API (google-adk 2.x).
# ParallelAgent is deprecated in favor of Workflow, and the cell above never ran: its
# model is retired (gemini-2.0-flash-exp -> 404) and its Runner block is all commented out.
#
# Fan-out / fan-in in a Workflow: a tuple of nodes runs concurrently, a JoinNode waits for
# all of them, and hands the next node a dict keyed by upstream node name. That join dict
# is what replaces the book's output_key + session.state hand-off.

from google.adk import Agent, Workflow
from google.adk.runners import InMemoryRunner
from google.adk.workflow import JoinNode
from google.genai import types

GEMINI_MODEL = "gemini-flash-latest"

# The two nodes that run in parallel. As the book notes, a real version would give these
# agents tools; with instructions alone the model simply makes the data up.
weather_fetcher = Agent(
    name="weather_fetcher",
    model=GEMINI_MODEL,
    instruction="Fetch the weather for the given location and return only the weather report.",
)

news_fetcher = Agent(
    name="news_fetcher",
    model=GEMINI_MODEL,
    instruction="Fetch the top news story for the given topic and return only that story.",
)

# The synchronization barrier: waits for both fetchers, then packages their outputs.
gather = JoinNode(name="gather_data")


def report(node_input: dict[str, str]) -> str:
    """Receives the join dict, keyed by the name of each upstream node."""
    print("\n--- Gathered Data ---")
    print(f"Weather: {node_input['weather_fetcher']}")
    print(f"News: {node_input['news_fetcher']}")
    return "Gathered weather and news."


# START fans out to both fetchers, they fan back in through the join node, and the
# joined dict goes to report.
data_gatherer = Workflow(
    name="data_gatherer",
    edges=[("START", (weather_fetcher, news_fetcher), gather, report)],
)

# --- Run the workflow ---
APP_NAME = "parallel_gatherer_app"
USER_ID = "user_1"

runner = InMemoryRunner(agent=data_gatherer, app_name=APP_NAME)
session = await runner.session_service.create_session(app_name=APP_NAME, user_id=USER_ID)

message = types.Content(
    role="user",
    parts=[types.Part(text="Get weather for Mountain View and news on technology")],
)

async for event in runner.run_async(
    user_id=USER_ID, session_id=session.id, new_message=message
):
    if event.is_final_response() and event.content and event.content.parts:
        text = "".join(part.text for part in event.content.parts if part.text)
        if text:
            print(f"[{event.author}] {text}")
```

## ADK Gemini Sequential

```python
from google.adk.agents import SequentialAgent, Agent

# This agent's output will be saved to session.state["data"]
step1 = Agent(name="Step1_Fetch", output_key="data")

# This agent will use the data from the previous step.
# We instruct it on how to find and use this data.
step2 = Agent(
    name="Step2_Process",
    instruction="Analyze the information found in state['data'] and provide a summary."
)

pipeline = SequentialAgent(
    name="MyPipeline",
    sub_agents=[step1, step2]
)

# When the pipeline is run with an initial input, Step1 will execute,
# its response will be stored in session.state["data"], and then
# Step2 will execute, using the information from the state as instructed.
```

### ADK Gemini Sequential with Workflow

This is the simplest of the three migrations in this notebook. A sequential pipeline is just a chain of edges:

| `SequentialAgent` (above) | `Workflow` (below) |
|---|---|
| `sub_agents=[step1, step2]` | `edges=[("START", step1, step2)]` |
| step1 writes `output_key="data"`; step2's instruction reads `state['data']` | the edge hands step1's output straight to step2 as its input |

But the deprecation is not the worst thing in the cell above. `step1 = Agent(name="Step1_Fetch", output_key="data")` has **no model and no instruction**. It constructs without complaint and would fail the instant anything invoked it — and, as with the coordinator and the parallel gatherer, nothing does.

That makes three cells in this notebook with the same shape of fault, which is worth saying out loud to a class: in ADK, constructing an agent validates its *fields*, not its *viability*. A model name is not resolved until the first call. A missing model is not noticed until inference is attempted. A method body does not run until something drives it. **A cell that only builds objects and asserts their relationships has told you nothing about whether the thing works.**

With this cell the set is complete. One graph API now replaces all three deprecated orchestration agents:

| Pattern | Deprecated agent | `Workflow` form |
|---|---|---|
| Sequential | `SequentialAgent` | a chain: `("START", a, b)` |
| Parallel | `ParallelAgent` | a tuple plus a `JoinNode`: `("START", (a, b), join, report)` |
| Loop | `LoopAgent` | a conditional edge pointing backwards: `(check, {"continue": a, "done": report})` |

The through-line: these were three different *agent types* because orchestration used to live in the agent. In a `Workflow` orchestration lives in the graph, so all three become different *edge shapes* instead. That is the migration worth remembering, rather than any individual signature.

```python
# Same two-step pipeline, written with the graph-based Workflow API (google-adk 2.x).
# SequentialAgent is deprecated in favor of Workflow.
#
# The cell above could never have run even before the deprecation: Step1_Fetch is an
# Agent with no model and no instruction. Constructing it succeeds, so the fault stays
# invisible until something actually invokes it - and nothing here does.

from google.adk import Agent, Workflow
from google.adk.runners import InMemoryRunner
from google.genai import types

GEMINI_MODEL = "gemini-flash-latest"

step1 = Agent(
    name="Step1_Fetch",
    model=GEMINI_MODEL,
    instruction="List three short, concrete facts about the topic the user names.",
)

step2 = Agent(
    name="Step2_Process",
    model=GEMINI_MODEL,
    # The book said "Analyze the information found in state['data']". No state key is
    # needed now: the edge hands step1's output straight to step2 as its input.
    instruction="Analyze the information you receive and reply with a one-sentence summary.",
)

# A sequential pipeline is just a chain of edges. The output of each node becomes the
# input of the next, which is what replaces output_key + session.state.
pipeline = Workflow(
    name="MyPipeline",
    edges=[("START", step1, step2)],
)

# --- Run the pipeline ---
APP_NAME = "sequential_pipeline_app"
USER_ID = "user_1"

runner = InMemoryRunner(agent=pipeline, app_name=APP_NAME)
session = await runner.session_service.create_session(app_name=APP_NAME, user_id=USER_ID)

message = types.Content(
    role="user",
    parts=[types.Part(text="The Voyager 1 spacecraft.")],
)

async for event in runner.run_async(
    user_id=USER_ID, session_id=session.id, new_message=message
):
    if event.is_final_response() and event.content and event.content.parts:
        text = "".join(part.text for part in event.content.parts if part.text)
        if text:
            print(f"--- {event.author} ---")
            print(text)
            print()
```
